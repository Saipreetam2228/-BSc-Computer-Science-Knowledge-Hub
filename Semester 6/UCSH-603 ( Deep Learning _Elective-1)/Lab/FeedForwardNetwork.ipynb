```python
class FFNetwork:
  
  def __init__(self):
    self.w1= np.random.randn()
    self.w2= np.random.randn()
    self.w3= np.random.randn()
    self.w4= np.random.randn()
    self.w5= np.random.randn()
    self.w6= np.random.randn()
    self.b1= np.random.randn()
    self.b2= np.random.randn()
    self.b3= np.random.randn()

  def sigmoid(self, x):
    return 1/(1 + np.exp(-x))

  def model(self, x):
    x1, x2= x
    a1= self.w1*x1 + self.w2*x2 + self.b1
    a2= self.w3*x1 + self.w4*x2 + self.b2
    h1= self.sigmoid(a1)
    h2= self.sigmoid(a2)
    a3= self.w5*h1 + self.w6*h2 + self.b3
    yhat= self.sigmoid(a3)
    return a1, h1, a2, h2, a3, yhat
  
  def predict(self, X):
    y= []
    for x in X:
      _,_,_,_,_,result= self.model(x)
      y.append(result)
    return np.array(y)
  
  def fit(self, X, Y, epochs=1, lr= 0.1):
    for _ in range(epochs):
        dw1= 0
        dw2= 0
        dw3= 0
        dw4= 0
        dw5= 0
        dw6= 0
        db1= 0
        db2= 0
        db3= 0
        
        for (x,y) in zip(X,Y):
            a1, h1, a2, h2, a3, yhat= self.model(x)
            x1, x2= x
            dw5+= (yhat - y) * yhat * (1 - yhat) * h1
            dw6+= (yhat - y) * yhat * (1 - yhat) * h2
            db3+= (yhat - y) * yhat * (1 - yhat)
            
            dw3+= (yhat - y) * yhat * (1 - yhat) * self.w6 * h2 * (1 - h2) * x1
            dw4+= (yhat - y) * yhat * (1 - yhat) * self.w6 * h2 * (1 - h2) * x2
            db2+= (yhat - y) * yhat * (1 - yhat) * self.w6 * h2 * (1 - h2)
            
            dw1+= (yhat - y) * yhat * (1 - yhat) * self.w5 * h1 * (1 - h1) * x1
            dw2+= (yhat - y) * yhat * (1 - yhat) * self.w5 * h1 * (1 - h1) * x2
            db1+= (yhat - y) * yhat * (1 - yhat) * self.w5 * h1 * (1 - h1)
            
        self.w1= self.w1 - lr*dw1
        self.w2= self.w2 - lr*dw2
        self.w3= self.w3 - lr*dw3
        self.w4= self.w4 - lr*dw4
        self.w5= self.w5 - lr*dw5
        self.w6= self.w6 - lr*dw6
        self.b1= self.b1 - lr*db1
        self.b2= self.b2 - lr*db2
        self.b3= self.b3 - lr*db3
        
```
