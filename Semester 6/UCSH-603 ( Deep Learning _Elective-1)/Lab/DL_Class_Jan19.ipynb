```python
# MP Neuron
theta= 0
for theta in range(X.shape[1] + 1):
    correct= 0
    for x, y in zip(X_binarized_train.values, y_train.values):
        y_pred= (np.sum(x) >= theta)
        correct+= (y_pred == y)
    print(theta, correct)
```

```python
from sklearn.metrics import accuracy_score
class MPNeuron:
  
  def __init__(self):
    self.theta= 0
    
  def model(self, x):
    return 1 if (sum(x) >= self.theta) else 0
  
  def predict(self, X):
    y= []
    for x in X:
      result= self.model(x)
      y.append(result)
    return np.array(y)
  
  def fit(self, X, y):
    accuracy= {}
    
    for theta in range(X.shape[1] + 1):
      self.theta= theta
      y_pred= self.predict(X)
      accuracy[theta]= accuracy_score(y_pred, y)
      
    best_theta= max(accuracy, key= accuracy.get)
    self.theta = best_theta
    
    print('Optimal value of theta is', best_theta)
    print('Highest accuracy is', accuracy[best_theta])
```

```python
mpn= MPNeuron()
mpn.fit(X_binarized_train.values, y_train.values)
```

```python
X_binarized_test= X_test.apply(pd.cut, bins= 2, labels= [0,1])
y_test_pred= mpn.predict(X_binarized_test.values)
print(accuracy_score(y_test_pred, y_test.values))
```

```python
X= [0.5, 2.5]
Y= [0.2, 0.9]

def f(w,b,x):
    return 1.0 / (1 + np.exp(-(w*x+b)))

def error(w, b):
    err= 0.0
    for x,y in zip(X,Y):
        fx= f(w,b,x)
        err+= 0.5 * (fx - y) ** 2
    return err

def grad_b(w,b,x,y):
    fx= f(w,b,x)
    return (fx - y) * fx * (1 - fx)

def grad_w(w,b,x,y):
    fx= f(w,b,x)
    return (fx - y) * fx * (1 - fx) * x

def do_gradient_descent():
    epochs= 500
    w= 0
    b= 0
    eta= 1.0
    for i in range(epochs):
        dw= 0.0
        db= 0.0
        for x, y in zip(X,Y):
            dw+= grad_w(w,b,x,y)
            db+= grad_b(w,b,x,y)
        w= w - eta * dw
        b= b - eta * db
        print(w, b, error(w,b))

```

```python
class SigmoidNeuron:
  
  def __init__(self):
    self.w= None
    self.b= 0
    
  def model(self, x):
    return 1/(1 + np.exp(-(np.dot(self.w,x)+ self.b)))
  
  def predict(self, X):
    y= []
    for x in X:
      result= self.model(x)
      y.append(result)
    return np.array(y)
  
  def fit(self, X, Y, epochs=1, lr= 0.1):
#     self.w= np.random.randn(len(X[0]))
    self.w= np.zeros(len(X[0]))
    for _ in range(epochs):
        dw= 0
        db= 0
        for (x,y) in zip(X,Y):
            yhat= self.model(x)
            dw+= (yhat - y) * yhat * (1 - yhat) * x
            db+= (yhat - y) * yhat * (1 - yhat)
        self.w= self.w - lr*dw
        self.b= self.b - lr*db
        
```
