```python
#question 2(a.1)
#Get the average height of a team.
def avg_height(a):
  he=0
  length=len(a)
  for name,i in a.items():
    he=he+i['heigth']
  avhe=he/length
  return(avhe)
print(avg_height(team))

```

```python
#question 2(a.2)
#Get the average height of a team.
def avg_age(a):
  ag=0
  length=len(a)
  for name,i in a.items():
    ag=ag+i['age']
  avag=ag/length
  return(avag)
print(avg_age(team))

```

```python
#question 2(b)
# Get number of explosive/stamina/all-rounders you a team.(print in any format you desire)

def phy_count(a):
  exp=0
  sta=0
  all=0
  for name,i in a.items():
    if i['physique']=="explosive":
      exp=exp+1
    elif i['physique']=='stamina':
      sta+=1
    else:
      all+=1
    return(exp,sta,all)
print(phy_count(team))
print("explosive=",phy_count(team)[0])
print("stamina=",phy_count(team)[1])
print("all-round=",phy_count(team)[2])
```

```python
#question 2(c)
#Find the position that two players play(if any)
pos=[]
for name,i in team.items():
  pos.append(i['position'])
print(pos)
```

```python
#question 2(d)
#Find the position that no player plays(if any
for i in pos:
  if pos.count(i)==1:
    print(i)
    break
  else:
    print("no player plays")
```

```python
#question 2(e)
#Find the nature of the team. Based on number of realists and idealist if 70% of the team is of one side then conclude for that side
'''  Example
    if 7/10 are idealists then team is idealist.(same for realist)
    if 6/10 or 5/10 are idealist or realists then team is a "mix"   '''
def nature(a):
  real=0
  ide=0
  for name,i in a.items():
    if i['nature']=="realist":
      real=real+1
    else:
      ide=ide+1
  return(real,ide)


real, ide = nature(team)
print(nature(team))
if real>ide:
  print("realist")
elif real<ide:
  print("idealist")
else:
  print("mix")
```

```python
#question 2(f)
# Find the overall physique of the team. The physique that occurs most number of times is the physique of the team

def overall_physique(a):
  exp=0
  sta=0
  all=0
  for name,i in a.items():
    if i['physique']=="explosive":
      exp=exp+1
    elif i['physique']=='stamina':
      sta+=1
    else:
      all+=1
  if exp>sta and exp>all:
    return("explosive")
  elif sta>exp and sta>all:
    return("stamina")
  else:
    return("all-round")
print("overall physique of the team is:",overall_physique(team))
```

```python
#question 2(g)
#If a new player is to be added into the team then find the maximum age allowed so that the

def max_age(a):
  ag=0
  length=len(a)
  for name,i in a.items():
    ag=ag+i['age']
  avag=ag/length
  max_allowed_age = (24 * (length + 1)) - avag
  max_allowed_age = (24 * (length + 1)) - ag
  return max_allowed_age

print(max_age(team))
```

```python
#question 2(h)
# Write a print function to print in a clear way all the players with their respective positions in case coach wants to
#   review the lineup.(example:  <Name> Plays in position <position>" )

def print_players(a):
  for name,i in a.items():
    print(name,"plays in position",i['position'])
print_players(team)
```

```python
#question 2(i)
#Write a function to find all those athletes who are of a particular physique(take physique as argument to function)
def find_physique(a,physique):
  for name,i in a.items():
    if i['physique']==physique:
      print(name)
find_physique(team,"explosive")
```
