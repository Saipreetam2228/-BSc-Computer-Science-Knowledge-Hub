```python
import pandas as pd

# Create the DataFrame from the provided data
data = {
    'survived': [0, 1, 1, 1, 0],
    'pclass': [3, 1, 3, 1, 3],
    'sex': ['male', 'female', 'female', 'female', 'male'],
    'age': [22.0, 38.0, 26.0, 35.0, 35.0],
    'sibsp': [1, 1, 0, 1, 0],
    'parch': [0, 0, 0, 0, 0],
    'fare': [7.2500, 71.2833, 7.9250, 53.1000, 8.0500],
    'embarked': ['S', 'C', 'S', 'S', 'S'],
    'class': ['Third', 'First', 'Third', 'First', 'Third'],
    'who': ['man', 'woman', 'woman', 'woman', 'man'],
    'adult_male': [True, False, False, False, True],
    'deck': [None, 'C', None, 'C', None],
    'embark_town': ['Southampton', 'Cherbourg', 'Southampton', 'Southampton', 'Southampton'],
    'alive': ['no', 'yes', 'yes', 'yes', 'no'],
    'alone': [False, False, True, False, True]
}

df = pd.DataFrame(data)
df.head()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
tips=sns.load_dataset('tips')
tips.head()
```

```python
# Plot 1: Boxplot
plt.figure(figsize=(8,6))
sns.boxplot(x='day', y='total_bill', data=tips,palette= 'YlOrBr')
plt.title('Boxplot of Total Bill by Day')
plt.xlabel('Day of the Week')
plt.ylabel('Total Bill')
plt.show

```

```python
# Plot 2: Scatter plot
plt.figure(figsize=(8,6))
sns.scatterplot(x='total_bill', y='tip', data=tips, hue='day', palette='YlOrBr')
plt.title('Scatter Plot of Total Bill vs Tip by Day')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
```

```python
# Plot 3: KDE plot - Kernel Density Estimate of tips
plt.figure(figsize=(8,6))
sns.kdeplot(data=tips, x='tip', hue='day', fill=True, common_norm=False, palette='YlOrBr')
plt
```

```python
# Plot 5: Histogram - Distribution of Total Bill
plt.figure(figsize=(8,6))
sns.histplot(data=tips, x='total_bill', kde=True, palette='YlOrBr')
plt.title('Distribution of Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt
```

```python
# Plot 6: Pair plot - Relationships between variables in the dataset
plt.figure(figsize=(10,8))
sns.pairplot(data=tips, hue='day', palette='YlOrBr')
plt.suptitle('Pair Plot of Tips Dataset', y=1.02)
```

```python
# Joint plot (age vs fare)
sns.jointplot(x='age', y='fare', data=df, kind='scatter')
plt.xlabel('Age')
plt.ylabel('Fare')
```

```python
# Distplot (distribution of fare)
sns.histplot(df['fare'], kde=True)
plt.xlabel('Fare')
```

```python
# KDE plot (age distribution separated by survival)
sns.kdeplot(data=df, x='age', hue='survived', fill=True, common_norm=False)
plt.xlabel('Age')
```

```python
# Correlation matrix (on numerical data)
corr_matrix = df[['age', 'sibsp', 'parch', 'fare']].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')

```
